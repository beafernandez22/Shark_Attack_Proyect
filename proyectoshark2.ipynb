```python
import pandas as pd
url = "GSAF5 (1).xls"
df = pd.read_excel(url)
print(df.head())
```

```python
# Comprobar valores nulos en Country y State
df[["Country", "State"]].isnull().sum()
```

```python
# Comprobar cuántos valores diferentes hay
df[["Country", "State"]].nunique()
```

```python
# Ver los países que aparecen con más frecuencia
df["Country"].value_counts().head(30)
```

```python
# Ver los estados que aparecen con más frecuencia
df["State"].value_counts().head(30)
```

```python
# Limpiar espacios y unificar mayúsculas/minúsculas en Country

df["Country"] = (
    df["Country"]
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
    .str.upper()
)
```

```python
# Comprobar limpieza de Country

df["Country"].nunique()
```

```python
df["Country"].value_counts().head(30)
```

```python
# Limpiar espacios y unificar mayúsculas/minúsculas en State

df["State"] = (
    df["State"]
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
    .str.title()
)
```

```python
# Comprobar cuántos estados diferentes quedan

df["State"].nunique()
```

```python
# Comprobar los estados después de la limpieza

df["State"].value_counts().head(30)
```

```python
# Sustituir localizaciones desconocidas sin inventar información

df["Country"] = df["Country"].fillna("UNKNOWN")
df["State"] = df["State"].fillna("Unknown")
```

```python
# Comprobar que ya no quedan nulos

df[["Country", "State"]].isnull().sum()
```

```python
# Ver los países menos frecuentes para detectar posibles errores
df["Country"].value_counts().tail(30)
```

```python
# Buscar valores de Country que contienen caracteres sospechosos
suspicious_countries = df[
    df["Country"].str.contains(r"\?|/|&", regex=True, na=False)
]["Country"].value_counts()

suspicious_countries
```

```python
# Ver los estados/regiones menos frecuentes para detectar posibles errores
df["State"].value_counts().tail(30)
```

```python
# Comprobación final de Country y State

print("Nulos:")
print(df[["Country", "State"]].isnull().sum())

print("\nValores únicos:")
print(df[["Country", "State"]].nunique())

print("\nEjemplo de los datos limpios:")
print(df[["Country", "State"]].head(10))
```
