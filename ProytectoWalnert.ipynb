```python
#sumar valores nulos de todas las columnas
df.isna().sum()
```

```python
#calcular con la moda los valores unicos
moda_year = df['Year'].mode()[0]
```

```python
#rellenar con la moda estos valores nulos
df['Year'] = df['Year'].fillna(moda_year)
```

```python
#cambiar de float a entero toda la columna de Year
df['Year'] = df['Year'].astype(int)
```

```python
#ver los valores nulos ahora
df["Year"].isna().sum()
```

```python
#Categorizar los valores por años
bins = [1, 1800, 1900, 1910, 1920, 1930, 1940, 1950, 1960, 1970, 1980, 1990, 2000, 2010, 2020, 2030]
labels = ['1-1800', '1800-1900', '1900-1910', '1910-1920', '1920-1930', '1930-1940', '1940-1950', '1950-1960', '1960-1970', '1970-1980', '1980-1990', '1990-2000', '2000-2010', '2010-2020', '2020-2030']

df['rango_year'] = pd.cut(df['Year'], bins=bins, labels=labels)
```

```python
#ordenar los valores categorizados de menor a mayor
df['rango_year'].value_counts().sort_index()
```

```python
#sumar los valores nulos de Activity
df['Activity'].isna().sum()
```

```python
#ver el tipo de dato que se encuentra en la columna
df['Activity'].dtype
```

```python
#rellenar los valores nulos con Unknown
df["Activity"] = df["Activity"].fillna("Unknown")
```

```python
#ver los valores unicos 
df['Activity'].unique()
```

```python
#regex se usa para limpieza de texto
df["Activity"] = (
    df["Activity"]
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
    .str.upper()
)
```
